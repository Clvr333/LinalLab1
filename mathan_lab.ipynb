# **EASY**

$x_{n+1} = rx_n(1-x_n)$

$\forall n \in \mathbb{N}, \forall r \in (0; 1]$

$$0 < x_0 < 1 \Rightarrow 0 < x_n < 1$$

$Докажем\ с\ помощью\ метода\ математической\ индукции:$

$База: 0 < x_0 < 1 - верно$

$Предположение: 0 < x_k < 1$

$0 < x_k < 1 ⇒ 0 < 1 - x_k < 1 ⇒ 0 < x_k(1 - x_k) < 1$

$\left.
\begin{aligned}
  &0 < x_k \cdot (1 - x_k) < 1 \\
  &0 < r \leq 1
\end{aligned}
\right\}
⇒\; 0 < rx_0(1 - x_0) = x_{k+1}< 1 ⇒ $

$⇒ \forall n \in \mathbb{N}\quad0 < x_n < 1$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def logistic_map(x, r):
  return r * x * (1 - x)

def g(x, r):
  return r * x * (1 - x ** 2)
```

```python
x = np.linspace(0, 1, 1000)
r_values = [0.5, 1.5, 2.5, 3.5, 4.0]

plt.figure(figsize=(10, 7))

for r in r_values:
  y = logistic_map(x, r)
  plt.plot(x, y, label=f'r = {r}', linewidth=2)

plt.plot(x, x, 'k--', alpha=0.3, label='$x_n = x_{n-1}$')

plt.title('Зависимость $x_n$ от $x_{n-1}$')
plt.xlabel('$x_{n-1}$')
plt.ylabel('$x_n$')
plt.legend()
plt.grid(True, linestyle=':')
plt.xlim(0, 1)
plt.ylim(0, 1.1)

plt.show()
```

$Проанализируем\ функцию\ f(x) = rx(1-x):$

$f(x) = -rx^2 + rx$

$График\ представлят\ собой\ параболу\ с\ ветвями\ вниз\ и\ координатами\ вершины:$

$x_0 = \frac{-r}{-2r} = \frac{1}{2} = const \\
y_0 = -\frac{1}{4}r + \frac{1}{2}r = \frac{1}{4}r$

$Параметр\ r\ влияет\ на\ высоту\ вершины\ и\ на\ растяжение\ параболы\ по\ оси\ y:$

$С\ увеличением\ r\ координата\ y\ вершины\ параболы\ увеличивается,\ график\ растягивается$

```python
x = np.linspace(0, 1, 1000)
r_vals = [0.5, 1.0, 1.5, 2.0, 2.5]

plt.figure(figsize=(10, 7))
for r in r_vals:
  y = g(x, r)
  plt.plot(x, y, label=f'r = {r:.1f}', linewidth=2)

plt.plot(x, x, 'k--', alpha=0.5, label=r'$y = x$')
plt.title(r'$g(x) = rx(1 - x^2)$')
plt.xlabel(r'$x_{n-1}$')
plt.ylabel(r'$x_n$')
plt.xlim(0, 1)
plt.ylim(0, 1.1)
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

$Проанализируем\ функцию\ g(x) = rx(1-x^2):$

$График\ представлят\ собой\ кубическую\ параболу\$

$Исследуем\ его\ на\ отрезке\ [0; 1],\ для\ этого\ найдём\ производную:$

$g'(x) = r(1 - 3x^2)$

$Нули\ производной:\ \frac{1}{\sqrt{3}}, -\frac{1}{\sqrt{3}}$

$Наибольшее\ значение\ принимает\ в\ точке\ x_m = \frac{1}{\sqrt{3}}$

$g_{max} = \frac{2r}{3\sqrt{3}}$

$Параметр\ r\ влияет\ на\ величину\ точки\ максимума\ и\ на\ растяжение\ графика\ по\ оси\ y:$

$С\ увеличением\ r\ растёт\ точка\ максимума,\ график\ растягивается$

Поведение обоих графиков схожее: оба имеют одну точку максимума (на рассматриваемом промежутке), обращаются в ноль в точках $(0; 0)$ и $(1; 0)$. Однако точка максимума второго немного сдвинута вправо из-за свойств кубической параболы

# **NORMAL**

$Рассмотрим\ f(x)\ =\ rx(1-x)\ на\ [0; 1]\ при\ 0 < r < 4:$

$Решим уравнение f(x) = x, чтобы найти неподвижные точки:$

$rx - rx^2 = x$

$x_1 = 0;\ x_2 = \frac{r-1}{r}$

$При\ 0 < r < 1:$

$x_2 < 0 \Rightarrow отображение\ имеет\ одну\ неподвижную\ точку\ x = 0$

$При\ r = 1:$

$x_2 = 0 = x_1 \Rightarrow отображение\ имеет\ одну\ неподвижную\ точку\ x = 0$

$При\ 1 < r < 4:$

$x_2 \in (0; 1) \Rightarrow\ отображение\ имеет\ две\ неподвижные точки\ x = 0\ и\ x = \frac{r-1}{r} $

$Максимальное\ количество\ точек - 2$


$x_0 \in (0; 1);\ r \in (0; 1]$

$x_{n+1} = rx_n(1 - x_n)$

$0 < x_n < 1 \Rightarrow 0 < 1 - x_n < 1$

$\left.
\begin{aligned}
  &r \leq 1 \\
  &1 - x_n < 1 \\
\end{aligned}
\right\} \Rightarrow rx_n(1 - x_n) = x_{n+1} < x_n \Rightarrow последовательность\ монотонно\ убывает$

$\left.
\begin{aligned}
  &0 < r\\
  &0 < 1 - x_n\\
\end{aligned}
\right\} \Rightarrow \forall n\ x_n >= 0$

$Получаем,\ что\ последовательность\ монотонно\ убывает\ и\ ограничена\ снизу \Rightarrow по\ теореме\ Вейерштрасса\ \exists \lim_{n \to \infty}{x_n}$

```python
r_values = [0.2, 0.5, 0.9, 1.0]
x0 = 0.8
n_iterations = 6

plt.figure(figsize=(12, 6))

for r in r_values:
  x_history = [x0]
  for _ in range(n_iterations):
    next_x = logistic_map(r, x_history[-1])
    x_history.append(next_x)

  plt.plot(x_history, marker='o', label=f'r = {r}')

plt.title('Визуализация последовательности $x_n$')
plt.xlabel('$n$')
plt.ylabel('$x_n$')
plt.grid(True, which='both', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

$f(x) = rx(1-x)$

$x_{n+1} = rx_n(1-x_n)$

$x_{2n} > x\ast;\ r \in (2; 3)$

$x\ast = \frac{r - 1}{r}$

$f'(x) = r - 2rx$

$f'(x\ast) = r - 2r\frac{r-1}{r} = 2 - r$

$-1 < f'(x*) < 0$

$Пусть\ g(x) = f(f(x))$

$g'(x\ast) = f'(f(x\ast)) \cdot f'(x\ast) = f(x\ast) \cdot f(x\ast) = (2 - r)^2$

$0 < g'(x\ast) < 1 \Rightarrow x\ast < x_{2n + 2} < x_{2n} \Rightarrow \{x_{2n}\}\ монотонно\ убывает$

$x_{2n+1} < x_{2n+3} < x\ast \Rightarrow \{x_{2n+1}\}\ монотонно\ возрастает$

```python
r = 2.7
x_star = (r - 1) / r

x_curr = 0.9
n_steps = 20

history = [x_curr]
for _ in range(n_steps):
  x_curr = r * x_curr * (1 - x_curr)
  history.append(x_curr)

steps = np.arange(len(history))
history = np.array(history)

evens = steps[::2]
evens_vals = history[::2]

odds = steps[1::2]
odds_vals = history[1::2]

plt.figure(figsize=(10, 6))

plt.axhline(y=x_star, color='black', linestyle='--', alpha=0.6, label=f'Неподвижная точка $x^*$ ≈ {x_star:.3f}')

plt.plot(evens, evens_vals, 'ro-', label=r'Подпоследовательность $\{x_{2n}\}$ (убывает)')

plt.plot(odds, odds_vals, 'bo-', label=r'Подпоследовательность $\{x_{2n+1}\}$ (возрастает)')

plt.title(f'Подпоследовательности логистического отображения (r={r})', fontsize=12)
plt.xlabel('$n$')
plt.ylabel('$x_n$')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.show()
```

$g(x) = rx(1 - x^2)$

$x_{n+1} = rx_n(1 - x_n^2)$

$0 < x_n < 1$

$rx(1 - x^2) = x$

$x\ast = \sqrt{\frac{r - 1}{r}}$

$Чтобы\ x_n\ монотонно\ сходилась\ к\ 0,\ необходимо:$

$1)\ |g'(0)| \leq 1$

$2)\ 0 < x_{n+1} < x_n$

$g'(0) = r - 3rx^2 = r \Rightarrow |r| <= 1$

$r > 0\ для\ монотонности:$

$rx(1 - x^2) < x$

$r(1 - x^2) < 1 - верно\ \forall x \in (0; 1),\ т. к.\ (1 - x^2) < 1$

$\Rightarrow r \in (0; 1]$

```python
x0=0.5
n_steps=30
r_values=[0.7, 1.0, 1.5, 2.3]
plt.figure(figsize=(10, 6))

for r in r_values:
  x_history = [x0]
  for _ in range(n_steps):
    x_history.append(g(x_history[-1], r))

  plt.plot(range(len(x_history)), x_history, 'o-', label=f'r = {r}')

plt.axhline(0, color='black', linewidth=1, alpha=0.5)
plt.title(f'Зависимость $x_n$ от $n$ при $x_0 = {x0}$')
plt.xlabel('$n$')
plt.ylabel('$x_n$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

# **HARD**

$В\ интервале\ существуют\ только\ циклы,\ длина\ которых\ равна\ степень\ двойки$

```python
def logistic_map2(r, x0, n):
  x = np.zeros(n)
  x[0] = x0
  for i in range(1, n):
    x[i] = r * x[i-1] * (1 - x[i-1])
  return x

r_values = [
    (3.2, "Цикл m=2"),
    (3.5, "Цикл m=4"),
    (3.55, "Цикл m=8"),
    (3.567, "Цикл m=16")
]

n_steps = 60
x0 = 0.5

fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)
fig.suptitle('Анализ отображения при r > 3', fontsize=16)

for i, (r, label) in enumerate(r_values):
  x = logistic_map2(r, x0, 200)
  tail = x[-n_steps:]

  axes[i].plot(range(n_steps), tail, 'o-', markersize=4, label=f'r = {r}')
  axes[i].set_title(label, loc='right', fontsize=10, color='red')
  axes[i].set_ylabel('$x_n$')
  axes[i].grid(True, alpha=0.3)
  axes[i].legend(loc='upper left')

plt.xlabel('$n$')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
def plot_lamerey(r, x0, n_steps=50):
  x_range = np.linspace(0, 1, 500)
  y_range = logistic_map(x_range, r)

  plt.figure(figsize=(8, 8))
  plt.plot(x_range, y_range, 'r', label=f'$y = {r}x(1-x)$')
  plt.plot(x_range, x_range, 'k--', alpha=0.5, label='$y = x$')

  x_curr = x0
  for _ in range(n_steps):
    x_next = logistic_map(x_curr, r)

    plt.plot([x_curr, x_curr], [x_curr, x_next], 'b', linewidth=0.8, alpha=0.6)
    plt.plot([x_curr, x_next], [x_next, x_next], 'b', linewidth=0.8, alpha=0.6)

    x_curr = x_next

  plt.title(f'Лестница Ламерея для r = {r}')
  plt.xlabel('$x_n$')
  plt.ylabel('$x_{n+1}$')
  plt.legend()
  plt.grid(True, alpha=0.3)
  plt.xlim(0, 1)
  plt.ylim(0, 1)
  plt.show()

plot_lamerey(r=3.2, x0=0.1)
plot_lamerey(r=3.5, x0=0.1)
plot_lamerey(r=3.55, x0=0.1)
plot_lamerey(r=3.567, x0=0.1)
```
